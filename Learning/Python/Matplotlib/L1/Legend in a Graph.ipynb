```python
import matplotlib.pyplot as plt
import numpy as np 
```

```python
a = np.arange(5)
b = [2,4,6,8,10]
c = [5,6,7,8,9]
```

```python
fig = plt.figure()
ax = plt.subplot()
```

```python
ax.plot(a, b, 'k--', label='Frequency')
ax.plot(a, c, 'k:', label='Periods')
```

```python
ax.legend()
```

```python
plt.title("Frequency of Signal")
plt.show
```
