```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = {
    "Cricket_Bed" : ["SG","BDM", "SS", "GM", "SPARTAN"],
    "MRP": [2000,2000,2400,2800, 2700],
    "Weight_Grams" : [1100,1200,1300,1350, 1600]
}
```

```python
dataFrame = pd.DataFrame(data)
```

```python
plt.xlabel("Bat Price USD")
plt.ylabel("Bat Weight Gram ")
plt.title("Bat Price depend on the bad")
plt.plot(dataFrame["MRP"], dataFrame["Weight_Grams"])
```

```python
plt.xlabel("Bat Price USD")
plt.ylabel("Bat Weight Gram ")
plt.title("Bat Price depend on the bad", loc="left")
plt.plot(dataFrame["MRP"], dataFrame["Weight_Grams"])
```
