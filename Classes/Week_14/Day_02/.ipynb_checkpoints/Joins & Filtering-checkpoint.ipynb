## Full/Outer Join

```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
```

```python
left_table = {'Date': ['01-01-2020', '02-01-2020', '03-01-2020', '04-01-2020','04-01-2025' ],
              'CountryID': [1,1,3,2,5],
              'Units': [40, 25, 30, 35, 55]}

left_table_df = pd.DataFrame(left_table)
left_table_df
```

```python
right_table = {'ID': [1,2,3,4],
              'Country': ['USA', 'Canada', 'Panama', 'Spain']}

right_table_df = pd.DataFrame(right_table)
right_table_df
```

```python
outer_join = left_table_df.merge(right_table_df, left_on='CountryID', right_on='ID', how='outer')
outer_join
```

```python
outer_join = left_table_df.merge(right_table_df, left_on='CountryID', right_on='ID', how='outer').drop('ID', axis=1)
outer_join
```

## Right Anti Join

```python
left_table = {'Data': ['01-01-2020', '02-01-2020', '03-01-2020', '04-01-2020' ],
            'CountryID': [1,1,3,2],
            'Units': [40,25,30,35]
           }
left_table_df = pd.DataFrame(left_table)
left_table_df
```

```python
right_table = {'ID': [3,4],
                'Country': ['Panama', 'Spain']
              }
right_table_df = pd.DataFrame(right_table)
right_table_df
```

```python
right_join_df = left_table_df.merge(right_table_df, left_on='CountryID', right_on='ID', how='right')
right_join_df
```

```python
right_join_df = left_table_df.merge(right_table_df, left_on='CountryID', right_on='ID', how='right').drop('ID', axis=1)
right_join_df
```

```python
right_join_df = left_table_df.merge(right_table_df, left_on='CountryID', right_on='ID', how='left', indicator=True).drop('ID', axis=1)
right_join_df
```

```python
right_join_df.loc[right_join_df['_merge'] == 'right_only']
```
